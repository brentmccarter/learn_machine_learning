# Simple Matplotlib charts
*  Intended for Visualization

```python
import matplotlib.pyplot as plt
import numpy as np

```

## 2D

```python
resultant = True

vectors = [[4,4],[5,4]]
array = np.array(vectors)

for vector in array:
    plt.quiver(0, 0, vector[0], vector[1], angles='xy', scale_units='xy', scale=1, label=f"V: {vector}", color = "blue")

# vector + vector
if resultant:
    sum = np.array([0,0])
    for vector in array:
        sum += vector
    plt.quiver(0, 0, sum[0], sum[1], angles='xy', scale_units='xy', scale=1, label=f"R: {sum}", color = "red")

# Set limits based on all vectors including sum
all_points = np.vstack([array, sum.reshape(1, -1)])
min_val = np.min(all_points, axis=0)
max_val = np.max(all_points, axis=0)

plt.xlim(min_val[0] if min_val[0] < 0 else 0, max_val[0] + 1)
plt.ylim(min_val[1] if min_val[1] < 0 else 0, max_val[1] + 1)
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```
