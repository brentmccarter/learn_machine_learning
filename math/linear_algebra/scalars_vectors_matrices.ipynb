# Scalars

A **scalar** is a quantity described only by its **magnitude** (size).

- A scalar is a single numerical value.
- Scalars do **not** have a direction.
- Scalars can be used to scale (multiply) vectors.

**Examples**
- 5
- -2
- 3.14
- Temperature
- Mass
- Time

---

# Vectors

## What is a Vector?

A **vector** is a quantity with both **magnitude** and **direction**.

A vector can be thought of as:

- An arrow pointing in space.
- An ordered list of numbers representing coordinates or features.

> "When thinking of vectors as an arrow in space, it is almost always rooted at the origin."
>
> — *3Blue1Brown*

Every vector corresponds to exactly one coordinate list, and every coordinate list represents exactly one vector.

Examples:

```text
(3,4)

[3,4]

⟨3,4⟩
```

These are simply different notations for the same vector.

---

## Vector Coordinates

Vector coordinates describe how to construct a vector using a coordinate system.

The standard basis vectors are

```text
î = (1,0)

ĵ = (0,1)
```

or equivalently

```text
e₁ = (1,0)

e₂ = (0,1)
```

Every coordinate tells you how much of each basis vector to use.

Example:

```text
a = (3,4)

a = 3î + 4ĵ
```

This means:

- Move 3 units in the x-direction.
- Move 4 units in the y-direction.

# Linear Combinations

A **linear combination** is formed by

1. Multiplying vectors by scalars.
2. Adding the resulting vectors.

Example:

```text
v₁ = (1,0)

v₂ = (0,1)

3v₁ + 2v₂

=

(3,2)
```

Another example:

```text
2(1,3) + (-1)(2,1)

=

(2,6) + (-2,-1)

=

(0,5)
```

Nearly every concept in linear algebra is built upon linear combinations.

---

## Span

The **span** of a set of vectors is the collection of **every possible linear combination** of those vectors.

Examples:

One vector:

```text
Span{(1,0)}
```

Produces every vector on the x-axis.

Two independent vectors:

```text
Span{(1,0),(0,1)}
```

Produces every point in the plane.

Three independent vectors in 3D:

```text
Span{(1,0,0),(0,1,0),(0,0,1)}
```

Produces all of 3-dimensional space.

Each vector adds a dimension to the span


---

## Linear Dependence

Vectors are **linearly dependent** if one vector can be written as a linear combination of the others.

In other words, at least one vector is redundant.

Example:

```text
v₁ = (1,2)

v₂ = (2,4)

v₂ = 2v₁
```

Removing one of these vectors does **not** change the span.

Another example:

```text
v₁ = (1,0)

v₂ = (0,1)

v₃ = (1,1)

v₃ = v₁ + v₂
```

Again, one vector is unnecessary.

---

## Linear Independence

A set of vectors is **linearly independent** if **no vector can be written as a linear combination of the others**.

Equivalently,

- Every vector contributes a new direction to the span.
- Removing any vector reduces the span.

Example:

```text
(1,0)

(0,1)
```

These vectors are linearly independent because neither can be built from the other. Each vector adds another dimension to the span.

---

## Basis

A **basis** is a set of vectors that

- Is linearly independent.
- Spans the entire vector space.

Every vector in the space can be written as **one unique linear combination** of the basis vectors.

Example (standard basis of ℝ²):

```text
(1,0)

(0,1)
```

Example:

```text
(5,2)

=

5(1,0)

+

2(0,1)
```

# Vector Magnitude (Length)

The magnitude (or norm) of a vector measures its length.

Formula:

```text
||(x,y)||

=

√(x²+y²)
```

Example:

```text
||(3,4)||

=

√(9+16)

=

5
```

---

# Unit Vectors

A **unit vector** has magnitude 1.

To convert any vector into a unit vector:

```text
u

=

v / ||v||
```

Example:

```text
v=(3,4)

||v||=5

u=(3/5,4/5)
```

# Vector Addition & Subtraction

## Addition

Add corresponding components.

```text
(x₁,y₁)

+

(x₂,y₂)

=

(x₁+x₂,

y₁+y₂)
```

Example:

```text
(2,3)

+

(4,1)

=

(6,4)
```

---

## Subtraction

Subtract corresponding components.

```text
(x₁,y₁)

-

(x₂,y₂)

=

(x₁-x₂,

y₁-y₂)
```

Example:

```text
(5,4)

-

(2,1)

=

(3,3)
```

---

## Geometric Interpretation

### Addition

Place the tail of the second vector at the head of the first.

The resulting vector extends from the original starting point to the final endpoint.

### Subtraction

Subtracting a vector is equivalent to adding its negative.

```text
a - b

=

a + (-b)
```

# Vector Multiplication

## Scalar Multiplication

Multiply every component by the scalar.

```text
k(x,y)

=

(kx,ky)
```

Properties:

- Changes magnitude.
- Preserves direction for positive scalars.
- Reverses direction for negative scalars.

Example:

```text
3(2,1)

=

(6,3)
```

---

## Dot Product

Multiply corresponding components and add the results.

Formula:

```text
(x₁,y₁) · (x₂,y₂)

=

x₁x₂ + y₁y₂
```

The result is always a **scalar**.

### Geometric Formula

```text
a·b

=

||a||||b||cos(θ)
```

where θ is the angle between the vectors.

### Uses

- Finding angles
- Measuring similarity
- Computing projections
- Determining orthogonality

Example:

```text
(1,2) · (3,4)

=

1·3 + 2·4

=

11
```

Orthogonal vectors satisfy

```text
a·b = 0
```

---

## Cross Product (3D)

The cross product is only defined in **three dimensions**.

It produces a vector that is perpendicular to both input vectors.

```text
a × b
```

Magnitude:

```text
||a×b||

=

||a||||b||sin(θ)
```

Properties:

- Produces a vector.
- Is perpendicular to both inputs.
- Direction follows the **right-hand rule**.

# Matrices

A **matrix** is a rectangular array of numbers.

Matrices can be viewed as

- Linear transformations
- Functions that map one vector to another
- Collections of basis vectors

A matrix takes an input vector and produces an output vector.

```text
Ax=b
```

where

- **A** is a matrix.
- **x** is the input vector.
- **b** is the output vector.

Matrices are commonly used to

- Rotate vectors
- Stretch vectors
- Reflect vectors
- Shear vectors
- Change coordinate systems
- Solve systems of linear equations

## Matrix as Column Vectors

Every matrix can also be viewed as a collection of column vectors.

Example:

```text
⎡1 3⎤
⎣2 4⎦
```

has columns

```text
(1,2)

(3,4)
```

Applying the matrix to a vector forms a linear combination of its columns.

---

### Matrix Multiplication

Matrix multiplication composes two linear transformations.

If

```text
A

transforms

x

into

y
```

and

```text
B

transforms

y

into

z
```

then

```text
BA
```

performs both transformations in sequence.

**Order matters.**

In general,

```text
AB ≠ BA
```
